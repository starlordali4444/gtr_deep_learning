    Where: L = Loss, ŷ = Output, z = Weighted Sum, w = Weight

```python
# Manual Backpropagation for Chai Sales Prediction
import numpy as np

def sigmoid(x): return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x): return x * (1 - x)

# Data: Temperature → High/Low chai sales
X = np.array([[15], [20], [25], [35]]) / 40  # Normalized
y = np.array([[1], [1], [0], [0]])  # 1=High, 0=Low sales

np.random.seed(42)
weight = np.random.randn(1, 1)
bias = np.random.randn(1)
learning_rate = 0.5
```

```python
for epoch in range(1000):
    # FORWARD PASS
    z = np.dot(X, weight) + bias
    predictions = sigmoid(z)
    error = y - predictions
    # BACKWARD PASS
    d_pred = error * sigmoid_derivative(predictions)
    # WEIGHT UPDATE
    weight += learning_rate * np.dot(X.T, d_pred)
    bias += learning_rate * np.sum(d_pred)
    print(f'Epoch => {epoch} \n Predictions => \n{predictions} \n Error => \n{error} | weight => {weight} | Bias => {bias}')
```

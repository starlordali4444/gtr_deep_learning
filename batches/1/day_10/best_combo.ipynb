```python
# =============================================================================
# IMPORTS
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, regularizers
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, Activation
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from tensorflow.keras.regularizers import l1, l2, l1_l2
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
```

```python
# Load MNIST dataset
print("\n📊 Loading MNIST Dataset...")
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()

# Preprocess - Normalize to 0-1 range
X_train_flat = X_train.reshape(-1, 784).astype('float32') / 255.0
X_test_flat = X_test.reshape(-1, 784).astype('float32') / 255.0

print(f"Training samples: {len(X_train_flat)}, Test samples: {len(X_test_flat)}")
```

```python

# =============================================================================
# BONUS: ULTIMATE REGULARIZED MODEL (Combining Everything!)
# =============================================================================
print("\n" + "=" * 60)
print("BONUS: ULTIMATE REGULARIZED MODEL")
print("Combining ALL Techniques!")
print("=" * 60)

def create_ultimate_model():
    """Create a model with ALL regularization techniques"""
    
    model = keras.Sequential([
        # Input Layer with light dropout
        layers.Input(shape=(784,)),
        layers.Dropout(0.1),  # Light dropout on input
        
        # Hidden Layer 1: Dense + BatchNorm + Dropout + L2
        layers.Dense(256, kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Dropout(0.3),
        
        # Hidden Layer 2
        layers.Dense(128, kernel_regularizer=regularizers.l2(0.001)),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Dropout(0.3),
        
        # Hidden Layer 3
        layers.Dense(64, kernel_regularizer=regularizers.l2(0.0005)),
        layers.BatchNormalization(),
        layers.Activation('relu'),
        layers.Dropout(0.2),
        
        # Output Layer
        layers.Dense(10, activation='softmax')
    ], name='Ultimate_Regularized_Model')
    
    return model

# Create and compile
print("\n🏆 Creating Ultimate Model...")
ultimate_model = create_ultimate_model()
ultimate_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Callbacks
callbacks = [
    EarlyStopping(monitor='val_loss', patience=15, restore_best_weights=True, verbose=1),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6, verbose=1)
]

# Train
print("\n🏋️ Training Ultimate Model...")
history_ultimate = ultimate_model.fit(
    X_train_flat, y_train,
    validation_split=0.2,
    epochs=50,
    batch_size=64,
    callbacks=callbacks,
    verbose=1
)

# Final evaluation
test_loss, test_acc = ultimate_model.evaluate(X_test_flat, y_test, verbose=0)
print(f"\n🎉 FINAL TEST RESULTS:")
print(f"   Test Accuracy: {test_acc*100:.2f}%")
print(f"   Test Loss: {test_loss:.4f}")


# =============================================================================
# SESSION SUMMARY
# =============================================================================
print("\n" + "=" * 60)
print("📝 DAY 9 SESSION SUMMARY")
print("=" * 60)

summary = """
┌─────────────────────────────────────────────────────────────┐
│ Technique          │ When to Use          │ Indian Analogy │
├─────────────────────────────────────────────────────────────┤
│ Dropout            │ Large networks       │ Kota Coaching  │
│ L1 Regularization  │ Feature selection    │ Wedding Cutter │
│ L2 Regularization  │ All features matter  │ Wedding Reducer│
│ Batch Normalization│ Deep networks        │ Chai Wala      │
│ Early Stopping     │ ALWAYS!              │ Gulab Jamun    │
│ Keras Tuner        │ Finding best config  │ Master Chef    │
└─────────────────────────────────────────────────────────────┘

🔑 KEY TAKEAWAYS:
1. Overfitting = memorizing without understanding (Ratta Maar)
2. Dropout = randomly deactivating neurons (Absent students)
3. L1 = sparse models (Cut wedding items)
4. L2 = shrink all weights (Reduce everything)
5. BatchNorm = standardize inputs (Chai Wala's secret)
6. Early Stopping = stop at the right time (Golden Gulab Jamun)
7. Combine techniques wisely - balance is key!

🎉 Congratulations! You've mastered Regularization & Hyperparameter Tuning!
📅 Next: Day 10 - Convolutional Neural Networks (CNN) 🖼️
"""

print(summary)

print("\n✅ All visualizations saved to outputs directory!")
print("=" * 60)
```

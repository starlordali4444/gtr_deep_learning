```python
# India's GDP growth over years
years = [2018, 2019, 2020, 2021, 2022, 2023]
gdp_growth = [6.8, 3.9, -6.6, 8.7, 7.2, 7.8]

# Create figure with 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# LINE CHART - for trends over time
ax1.plot(years, gdp_growth, marker='o', linewidth=2, color='#FF9933', markersize=8)
ax1.axhline(y=0, color='gray', linestyle='--', alpha=0.7)  # Zero line
ax1.set_xlabel('Year', fontsize=12)
ax1.set_ylabel('GDP Growth Rate (%)', fontsize=12)
ax1.set_title('India GDP Growth Rate (2018-2023)', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)

# Highlight COVID year
ax1.annotate('COVID Impact', xy=(2020, -6.6), xytext=(2020.5, -3),
            arrowprops=dict(arrowstyle='->', color='red'),
            fontsize=10, color='red')

# BAR CHART - for comparisons
cities = ['Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Kolkata']
population = [21, 19, 13, 11, 15]  # in millions

colors = ['#138808', '#FF9933', '#000080', '#FF9933', '#138808']  # Tricolor theme
ax2.bar(cities, population, color=colors, edgecolor='black')
ax2.set_xlabel('City', fontsize=12)
ax2.set_ylabel('Population (Millions)', fontsize=12)
ax2.set_title('Top 5 Indian Metro Cities by Population', fontsize=14, fontweight='bold')

# Add value labels on bars
for i, v in enumerate(population):
    ax2.text(i, v + 0.3, f'{v}M', ha='center', fontweight='bold')

plt.tight_layout()
plt.savefig('india_charts.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt

# PIE CHART - Budget allocation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Monthly household budget
categories = ['Rent', 'Food', 'Transport', 'Education', 'Savings', 'Entertainment']
amounts = [15000, 10000, 5000, 8000, 7000, 5000]
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7', '#DDA0DD']
explode = (0.05, 0, 0, 0, 0.05, 0)  # Highlight Rent and Savings

ax1.pie(amounts, labels=categories, autopct='%1.1f%%', colors=colors,
        explode=explode, shadow=True, startangle=90)
ax1.set_title('Monthly Household Budget\n(Total: ₹50,000)', fontsize=14, fontweight='bold')

# HISTOGRAM - Distribution of exam scores
import numpy as np
np.random.seed(42)
# Simulating marks of 200 students (bell curve around 60)
student_marks = np.random.normal(60, 15, 200)
student_marks = np.clip(student_marks, 0, 100)  # Keep between 0-100

ax2.hist(student_marks, bins=10, color='#3498db', edgecolor='black', alpha=0.7)
ax2.axvline(x=np.mean(student_marks), color='red', linestyle='--', 
            linewidth=2, label=f'Average: {np.mean(student_marks):.1f}')
ax2.axvline(x=35, color='orange', linestyle='--', 
            linewidth=2, label='Pass Mark: 35')
ax2.set_xlabel('Marks', fontsize=12)
ax2.set_ylabel('Number of Students', fontsize=12)
ax2.set_title('Class 10 Maths Marks Distribution', fontsize=14, fontweight='bold')
ax2.legend()

plt.tight_layout()
plt.savefig('pie_histogram.png', dpi=150)
plt.show()

# Calculate pass/fail
passed = sum(student_marks >= 35)
print(f"Students Passed: {passed} ({passed/2}%)")
print(f"Students Failed: {200-passed} ({(200-passed)/2}%)")
```

    Scatter Plot and Multiple Plots

```python
import matplotlib.pyplot as plt
import numpy as np

# Create a comprehensive analysis figure
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Data: Study hours vs Marks (100 students)
np.random.seed(42)
study_hours = np.random.uniform(1, 8, 100)
# Marks correlate with study hours + some randomness
marks = 40 + (study_hours * 7) + np.random.normal(0, 8, 100)
marks = np.clip(marks, 0, 100)

# SCATTER PLOT - Correlation between two variables
ax1 = axes[0, 0]
scatter = ax1.scatter(study_hours, marks, c=marks, cmap='RdYlGn', 
                       s=60, alpha=0.7, edgecolors='black')
ax1.set_xlabel('Daily Study Hours', fontsize=12)
ax1.set_ylabel('Exam Marks', fontsize=12)
ax1.set_title('Study Hours vs Marks\n(Each dot = 1 student)', fontsize=14, fontweight='bold')

# Add trend line
z = np.polyfit(study_hours, marks, 1)
p = np.poly1d(z)
ax1.plot(study_hours, p(study_hours), "r--", alpha=0.8, linewidth=2, label='Trend Line')
ax1.legend()
plt.colorbar(scatter, ax=ax1, label='Marks')

# MULTIPLE LINE CHART - Comparing trends
ax2 = axes[0, 1]
months = ['Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep']
sales_north = [45, 52, 48, 61, 55, 67]
sales_south = [38, 45, 52, 58, 62, 70]
sales_west = [42, 48, 44, 50, 53, 58]

ax2.plot(months, sales_north, marker='o', linewidth=2, label='North Zone', color='#e74c3c')
ax2.plot(months, sales_south, marker='s', linewidth=2, label='South Zone', color='#2ecc71')
ax2.plot(months, sales_west, marker='^', linewidth=2, label='West Zone', color='#3498db')
ax2.set_xlabel('Month', fontsize=12)
ax2.set_ylabel('Sales (₹ Lakhs)', fontsize=12)
ax2.set_title('Zone-wise Sales Comparison (FY 2024)', fontsize=14, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

# STACKED BAR CHART - Parts of a whole over categories
ax3 = axes[1, 0]
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
products = {
    'Electronics': [25, 30, 35, 40],
    'Clothing': [20, 22, 18, 25],
    'Groceries': [35, 32, 38, 30],
    'Others': [10, 12, 15, 18]
}

bottom = np.zeros(4)
colors = ['#3498db', '#e74c3c', '#2ecc71', '#9b59b6']
for i, (product, values) in enumerate(products.items()):
    ax3.bar(quarters, values, bottom=bottom, label=product, color=colors[i])
    bottom += values

ax3.set_xlabel('Quarter', fontsize=12)
ax3.set_ylabel('Sales (₹ Lakhs)', fontsize=12)
ax3.set_title('Quarterly Sales by Category', fontsize=14, fontweight='bold')
ax3.legend(loc='upper left')

# BOX PLOT - Distribution comparison
ax4 = axes[1, 1]
np.random.seed(42)
section_a = np.random.normal(65, 12, 40)
section_b = np.random.normal(70, 8, 40)
section_c = np.random.normal(55, 15, 40)

data = [section_a, section_b, section_c]
bp = ax4.boxplot(data, labels=['Section A', 'Section B', 'Section C'], patch_artist=True)

colors = ['#FF9999', '#99FF99', '#9999FF']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

ax4.set_ylabel('Marks', fontsize=12)
ax4.set_title('Section-wise Marks Comparison\n(Box Plot)', fontsize=14, fontweight='bold')
ax4.axhline(y=35, color='red', linestyle='--', alpha=0.7, label='Pass Mark')
ax4.legend()

plt.tight_layout()
plt.savefig('comprehensive_analysis.png', dpi=150)
plt.show()
```

    Seaborn (Beautiful Statistical Visualizations)
    Layman Definition
    Seaborn is Matplotlib's stylish younger sibling. It creates beautiful, professional-looking statistical charts with less code. It's especially good for understanding relationships and distributions in data.
    Indian Analogy 🇮🇳
    If Matplotlib is like a basic Maruti car - gets the job done, reliable, you control everything...
    Then Seaborn is like a top-model Honda/Hyundai - more features, looks better by default, automatic settings, but built on the same road (Matplotlib).
    Seaborn = "Matplotlib with automatic makeup and better fashion sense!"

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set the style - Seaborn's magic!
sns.set_style("whitegrid")
sns.set_palette("husl")

# Create sample data - Employee salaries
np.random.seed(42)
data = pd.DataFrame({
    'Salary': np.concatenate([
        np.random.normal(35000, 8000, 150),   # Junior employees
        np.random.normal(65000, 12000, 100),  # Mid-level
        np.random.normal(120000, 25000, 50)   # Senior
    ]),
    'Department': np.random.choice(['IT', 'Sales', 'HR', 'Finance'], 300),
    'Experience': np.random.randint(1, 20, 300)
})
data['Salary'] = data['Salary'].clip(20000, 200000)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# HISTOGRAM with KDE (Kernel Density Estimate)
ax1 = axes[0, 0]
sns.histplot(data=data, x='Salary', kde=True, ax=ax1, color='#3498db', bins=30)
ax1.set_title('Salary Distribution with Density Curve', fontsize=14, fontweight='bold')
ax1.set_xlabel('Monthly Salary (₹)')

# KDE PLOT - Smooth distribution by department
ax2 = axes[0, 1]
for dept in data['Department'].unique():
    dept_data = data[data['Department'] == dept]['Salary']
    sns.kdeplot(data=dept_data, ax=ax2, label=dept, linewidth=2)
ax2.set_title('Salary Distribution by Department', fontsize=14, fontweight='bold')
ax2.set_xlabel('Monthly Salary (₹)')
ax2.legend()

# BOX PLOT - Seaborn style
ax3 = axes[1, 0]
sns.boxplot(data=data, x='Department', y='Salary', ax=ax3, palette='Set2')
ax3.set_title('Department-wise Salary Comparison', fontsize=14, fontweight='bold')
ax3.set_ylabel('Monthly Salary (₹)')

# VIOLIN PLOT - Box plot + Distribution combined!
ax4 = axes[1, 1]
sns.violinplot(data=data, x='Department', y='Salary', ax=ax4, palette='muted')
ax4.set_title('Violin Plot: Salary by Department\n(Shows full distribution shape)', 
              fontsize=14, fontweight='bold')
ax4.set_ylabel('Monthly Salary (₹)')

plt.tight_layout()
plt.savefig('seaborn_distributions.png', dpi=150)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_style("whitegrid")

# Student performance data
np.random.seed(42)
n = 200
students = pd.DataFrame({
    'Study_Hours': np.random.uniform(1, 8, n),
    'Sleep_Hours': np.random.uniform(4, 9, n),
    'Attendance': np.random.uniform(50, 100, n),
    'Gender': np.random.choice(['Male', 'Female'], n),
    'Tuition': np.random.choice(['Yes', 'No'], n)
})

# Marks depend on multiple factors
students['Marks'] = (
    30 + 
    students['Study_Hours'] * 5 +
    students['Sleep_Hours'] * 2 +
    students['Attendance'] * 0.3 +
    np.random.normal(0, 8, n)
).clip(0, 100)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# SCATTER PLOT with regression line
ax1 = axes[0, 0]
sns.regplot(data=students, x='Study_Hours', y='Marks', ax=ax1, 
            scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
ax1.set_title('Study Hours vs Marks\n(with trend line)', fontsize=14, fontweight='bold')

# SCATTER with HUE - Adding third dimension with color
ax2 = axes[0, 1]
sns.scatterplot(data=students, x='Study_Hours', y='Marks', 
                hue='Tuition', style='Gender', ax=ax2, s=80)
ax2.set_title('Study Hours vs Marks\n(colored by Tuition, shaped by Gender)', 
              fontsize=14, fontweight='bold')

# CORRELATION HEATMAP - Most important!
ax3 = axes[1, 0]
numeric_cols = students.select_dtypes(include=[np.number])
correlation = numeric_cols.corr()
sns.heatmap(correlation, annot=True, cmap='RdYlGn', center=0, 
            ax=ax3, fmt='.2f', linewidths=0.5)
ax3.set_title('Correlation Heatmap\n(Green=Positive, Red=Negative)', 
              fontsize=14, fontweight='bold')

# PAIR PLOT concept shown as joint plot
ax4 = axes[1, 1]
sns.scatterplot(data=students, x='Attendance', y='Marks', hue='Study_Hours', 
                palette='viridis', ax=ax4, size='Study_Hours', sizes=(20, 200))
ax4.set_title('Attendance vs Marks\n(dot size & color = Study Hours)', 
              fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig('seaborn_relationships.png', dpi=150)
plt.show()

# Bonus: Create a pair plot (separate figure)
print("\n--- Creating Pair Plot (shows ALL relationships) ---")
pair_plot = sns.pairplot(students[['Study_Hours', 'Sleep_Hours', 'Attendance', 'Marks']], 
                          diag_kind='kde', corner=True)
pair_plot.fig.suptitle('Pair Plot: All Variables Against Each Other', y=1.02, fontsize=14)
plt.savefig('pair_plot.png', dpi=150, bbox_inches='tight')
plt.show()
```

    Categorical Plots (Comparing Groups)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

sns.set_style("whitegrid")

# E-commerce data - like Flipkart sales analysis
np.random.seed(42)
orders = pd.DataFrame({
    'Category': np.random.choice(['Electronics', 'Fashion', 'Home', 'Books'], 500),
    'Payment': np.random.choice(['UPI', 'Card', 'COD', 'NetBanking'], 500),
    'City_Tier': np.random.choice(['Tier 1', 'Tier 2', 'Tier 3'], 500),
    'Order_Value': np.random.exponential(2000, 500) + 500,
    'Rating': np.random.choice([1, 2, 3, 4, 5], 500, p=[0.05, 0.1, 0.2, 0.35, 0.3])
})
orders['Order_Value'] = orders['Order_Value'].clip(500, 15000).round(0)

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# COUNT PLOT - How many in each category
ax1 = axes[0, 0]
sns.countplot(data=orders, x='Category', hue='Payment', ax=ax1, palette='Set2')
ax1.set_title('Orders by Category & Payment Method', fontsize=14, fontweight='bold')
ax1.legend(title='Payment')
ax1.set_ylabel('Number of Orders')

# BAR PLOT with error bars - Average values comparison
ax2 = axes[0, 1]
sns.barplot(data=orders, x='Category', y='Order_Value', hue='City_Tier', 
            ax=ax2, palette='coolwarm', errorbar='sd')
ax2.set_title('Average Order Value by Category & City Tier\n(with standard deviation)', 
              fontsize=14, fontweight='bold')
ax2.set_ylabel('Order Value (₹)')
ax2.legend(title='City Tier')

# STRIP PLOT - Individual points (swarm)
ax3 = axes[1, 0]
sns.stripplot(data=orders, x='Category', y='Order_Value', hue='City_Tier',
              ax=ax3, dodge=True, alpha=0.6, size=4)
ax3.set_title('Individual Orders Distribution\n(Each dot = 1 order)', 
              fontsize=14, fontweight='bold')
ax3.set_ylabel('Order Value (₹)')
ax3.legend(title='City Tier', loc='upper right')

# POINT PLOT - Mean with confidence intervals
ax4 = axes[1, 1]
sns.pointplot(data=orders, x='Rating', y='Order_Value', hue='City_Tier',
              ax=ax4, palette='deep', markers=['o', 's', '^'], linestyles=['-', '--', ':'])
ax4.set_title('How Rating Relates to Order Value\n(by City Tier)', 
              fontsize=14, fontweight='bold')
ax4.set_ylabel('Average Order Value (₹)')
ax4.legend(title='City Tier')

plt.tight_layout()
plt.savefig('seaborn_categorical.png', dpi=150)
plt.show()

# Summary statistics
print("\n--- Quick Business Insights ---")
print("\nCategory-wise Average Order Value:")
print(orders.groupby('Category')['Order_Value'].mean().round(0))
print("\nPayment Method Popularity:")
print(orders['Payment'].value_counts())
```

```python
import pandas as pd

# Creating a DataFrame - like creating a shop's ledger
students = pd.DataFrame({
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Vikram'],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Mumbai'],
    'Maths': [85, 90, 78, 92, 88],
    'Science': [80, 85, 82, 88, 90],
    'English': [75, 88, 80, 85, 82]
})
```

```python
print("Our Class Data:")
print(students)
```

```python
print("\n--- Basic Information ---")
print(students.shape)      # (5, 5) - 5 students, 5 columns
print(students.columns)    # List of column names
print(students.dtypes)     # Data types of each column
```

```python
print("\n--- Quick Look ---")
print(students.head(2))    # First 2 rows
print(students.tail(2))    # Last 2 rows
print(students.sample(2))  # Random 2 rows
```

```python
print("\n--- Statistics ---")
print(students.describe()) # Count, mean, std, min, max for numbers
print("\n--- Info ---")
print(students.info()) # Count, mean, std, min, max for numbers
```

```python
import pandas as pd

# E-commerce orders data - like Flipkart/Amazon
orders = pd.DataFrame({
    'Order_ID': [101, 102, 103, 104, 105, 106],
    'Customer': ['Rahul', 'Priya', 'Amit', 'Sneha', 'Rahul', 'Priya'],
    'City': ['Delhi', 'Mumbai', 'Delhi', 'Chennai', 'Delhi', 'Mumbai'],
    'Product': ['Mobile', 'Laptop', 'Mobile', 'Tablet', 'Earphones', 'Mobile'],
    'Amount': [15000, 55000, 12000, 25000, 2000, 18000]
})
```

```python
print("All Orders:")
print(orders)
```

```python
# Select specific columns - like asking "sirf naam aur amount dikhao"
print("\n--- Only Customer and Amount ---")
print(orders[['Customer', 'Amount']])
```

```python
# Filter rows - "Delhi ke orders dikhao"
print("\n--- Delhi Orders Only ---")
delhi_orders = orders[orders['City'] == 'Delhi']
print(delhi_orders)
```

```python
# Multiple conditions - "Delhi mein 10000 se zyada ke orders"
print("\n--- Delhi Orders Above 10000 ---")
big_delhi = orders[(orders['City'] == 'Delhi') & (orders['Amount'] > 10000)]
print(big_delhi)
```

```python
# Find specific customer - "Rahul ke kitne orders hain?"
print("\n--- Rahul's Orders ---")
rahul_orders = orders[orders['Customer'] == 'Rahul']
print(rahul_orders)
print(f"Rahul ka total: ₹{rahul_orders['Amount'].sum()}")
```

```python
# Sales data for a retail chain
sales = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-03'],
    'Store': ['Delhi', 'Mumbai', 'Delhi', 'Mumbai', 'Delhi'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Electronics', 'Clothing'],
    'Sales': [50000, 30000, 45000, 60000, 25000],
    'Units': [5, 15, 4, 6, 12]
})
```

```python
print("Sales Data:")
print(sales)
```

```python
# Group by Store - "Har store ki total sale kitni?"
print("\n--- Store-wise Total Sales ---")
store_sales = sales.groupby('Store')['Sales'].sum()
print(store_sales)
```

```python
# Multiple aggregations - "Store-wise sum, average, count sab chahiye"
print("\n--- Store-wise Detailed Analysis ---")
store_analysis = sales.groupby('Store').agg({
    'Sales': ['sum', 'mean', 'count'],
    'Units': 'sum'
})
print(store_analysis)
```

```python
# Group by multiple columns
print("\n--- Store + Category wise Sales ---")
detailed = sales.groupby(['Store', 'Category'])['Sales'].sum()
print(detailed)
```

```python
# Pivot Table - like Excel pivot
print("\n--- Pivot Table (Store vs Category) ---")
pivot = sales.pivot_table(
    values='Sales', 
    index='Store', 
    columns='Category', 
    aggfunc='sum',
    fill_value=0
)
print(pivot)
```

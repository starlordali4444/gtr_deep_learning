        0   =>  start_range
        101 =>  ending_range
        5   =>  no of random values to generate
    )

```python
import numpy as np
```

```python
maths=np.random.randint(35,101,5)
maths
```

```python
science=np.random.randint(35,101,5)
science
```

```python
english=np.random.randint(35,101,5)
english
```

```python
maths.shape
```

```python
# Total Marks
total_marks=maths+science+english
total_marks
```

```python
# Calculate percentage
percentage=total_marks/3
percentage
```

```python
# Find statistics
print(f'Highest in Maths: {maths.max()}')
print(f'Highest in Maths: {np.max(maths)}')
```

```python
print(f'Average in english: {np.mean(english)}')
```

```python
# Marks of 4 students in 3 subjects (rows = students, columns = subjects)
# Columns: Maths, Science, English
class_marks = np.array([
    [85, 80, 75],   # Student 1: Rahul
    [90, 85, 88],   # Student 2: Priya
    [78, 82, 80],   # Student 3: Amit
    [92, 88, 85]    # Student 4: Sneha
])
```

```python
print("Full Class Data:")
print(class_marks)
```

```python
class_marks.shape
```

```python
# Get specific data
print("\nRahul's marks:", class_marks[0])        # First row
```

```python
print("Everyone's Maths:", class_marks[:, 0])    # First column
```

```python
print("Priya's Science:", class_marks[1, 1])     # Specific cell
```

    2D array 
        Row-Wise    =>  axis=1
        Column-Wise =>  axis=0

```python
# Student-wise total
student_totals = np.sum(class_marks, axis=1)  # axis=1 means row-wise
print("\nStudent totals:", student_totals)
```

```python
# Subject-wise total
subject_totals = np.sum(class_marks, axis=0)  # axis=0 means column-wise
print("Subject totals:", subject_totals)
```

    Loop
        A loop in Python is a control structure used to repeat a block of code multiple times until a condition is met or until you explicitly stop it.

    Used when you know the exact range or sequence to iterate over.
    
    for item in sequence:
        # repeat code


    Used when repetition depends on a condition, not a known sequence.
    while condition:
        # repeat code

    Find factorial of a number.

```python
n=5

factorial=1

for i in range(1,n+1):
    print(factorial,i)
    factorial=factorial*i       
print(f'The factorial of {n} is {factorial}')
```

```python
n=5
factorial=1
while n>0:
    print(factorial,n)
    factorial*=n
    n-=1
print(f'The factorial is {factorial}')
```

    Functions
        ✅ 1. What is a Function? (Definition)

            A function is a reusable block of code that performs a specific task.
            You define a function once, and you can call/use it multiple times.

        ✅ 2. Why Do We Use Functions?

        Avoid code repetition
        Improve readability
        Break big programs into smaller parts
        Make debugging easier
        Reusability
        Modularity

    def function_name(parameters):
        # block of code
        return value(optional)


    ✅ 4. Types of Functions

        1. Built-in Functions
        Examples: len(), print(), input(), sum(), max(), range()

        2. User-defined Functions
        Functions we write using def.

        3. Anonymous Functions (Lambda)
        Short functions written using lambda.

    ✅ 5. Parameters vs Arguments

        Parameters:
            Variables written inside function definition.
            Example:
            def add(a, b): # a, b are parameters

        Arguments:
            Values passed while calling the function.
            add(10, 20) # 10, 20 are arguments

```python
def add(a,b):
    return a+b  

add(3,5)
```

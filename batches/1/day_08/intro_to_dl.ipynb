## ANN Architecture: Building Blocks   

    The Three Layer Types

    INPUT LAYER
    •	Receives raw features from your dataset
    •	Number of neurons = Number of input features
    •	No computation happens here - just data entry point
    •	Example: For MNIST (28×28 images), input layer has 784 neurons (28×28 pixels flattened)

    HIDDEN LAYER(S)
    •	Where the 'magic' happens - patterns are learned here
    •	Can have multiple hidden layers (hence 'deep' learning)
    •	Number of neurons is a hyperparameter YOU choose
    •	Each layer extracts increasingly abstract features
    •	Rule of thumb: Start with neurons between input and output size, then tune
    
    OUTPUT LAYER
    •	Produces the final prediction
    •	Structure depends on the problem type:

    Problem Type	                    Output Neurons	    Activation
    Regression (predict number)	        1 neuron	        None (Linear)
    Binary Classification	            1 neuron	        Sigmoid
    Multi-class (N classes)	            N neurons	        Softmax

## Activation Functions 

#### Why Do We Need Activation Functions?
    Without activation functions, no matter how many layers you stack, the network would only learn linear relationships. Multiple linear transformations combined are still just a linear transformation. Activation functions introduce non-linearity, allowing the network to learn complex, non-linear patterns in data.


### The Four Essential Activation Functions

    ReLU (Rectified Linear Unit) - The Modern Default

    f(x) = max(0, x)  →  Output: 0 if x < 0, else x
    
    •	Range: [0, ∞) - Only positive values pass through
    •	Use: Default choice for hidden layers in most networks
    •	Pros: Computationally efficient, no vanishing gradient for positive values, sparse activation
    •	Cons: 'Dying ReLU' problem - neurons can get stuck at 0


    Sigmoid - The Probability Generator
    
    σ(x) = 1 / (1 + e⁻ˣ)  →  Squashes any input to range (0, 1)
    •	Range: (0, 1) - Perfect for probability interpretation
    •	Use: Output layer for binary classification
    •	Pros: Smooth gradient, output interpretable as probability
    •	Cons: Vanishing gradient at extremes, outputs not zero-centered


    Tanh (Hyperbolic Tangent) - The Zero-Centered Alternative

    tanh(x) = (eˣ - e⁻ˣ) / (eˣ + e⁻ˣ)  →  Squashes to range (-1, 1)
    •	Range: (-1, 1) - Zero-centered output
    •	Use: Hidden layers when zero-centered output helps (RNNs), or when negative outputs are meaningful
    •	Relation to Sigmoid: tanh(x) = 2·sigmoid(2x) - 1 (scaled and shifted sigmoid)

    Softmax - The Multi-Class Decision Maker

    softmax(xᵢ) = eˣⁱ / Σ(eˣʲ)  →  Converts to probability distribution (sum = 1)
    •	Range: (0, 1) for each output, sum of all outputs = 1
    •	Use: Output layer for multi-class classification (ONLY)
    •	Key Property: Outputs can be interpreted as probabilities for each class

    Forward Propagation: How Data Flows

    Forward propagation is the process of passing input data through the network, layer by layer, to generate an output prediction. Understanding this flow is essential before we learn how networks learn (backpropagation).

    Step-by-Step Forward Propagation
    1.	Input Layer: Receive input features x = [x₁, x₂, ..., xₙ]
    2.	Hidden Layer 1: z₁ = W₁·x + b₁, then a₁ = activation(z₁)
    3.	Hidden Layer 2: z₂ = W₂·a₁ + b₂, then a₂ = activation(z₂)
    4.	... Continue for all hidden layers
    5.	Output Layer: ŷ = output_activation(Wₒᵤₜ·aₗₐₛₜ + bₒᵤₜ)
